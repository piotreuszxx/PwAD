### Przykłady

```python
# importy potrzebne do przykładów
import numpy as np
import pandas as pd
```

```python
# wykres liniowy kwadratu liczby
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y, marker='o')
plt.title("Wykres liniowy – kwadraty liczb")
plt.xlabel("x")
plt.ylabel("x^2")
plt.grid(True)
plt.show()
```

```python
# wykres kołowy
labels = ['A', 'B', 'C', 'D']
values = [30, 15, 40, 15]

plt.figure(figsize=(6,6))
plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title("Wykres kołowy – Matplotlib")
plt.show()
```

```python
# wykres słupowy MatPlotLib
categories = ['Python', 'Java', 'C++', 'Go']
popularity = [45, 30, 15, 10]

plt.figure(figsize=(8,5))
plt.bar(categories, popularity)
plt.title("Popularność języków programowania")
plt.xlabel("Język")
plt.ylabel("Popularność (%)")
plt.show()
```

```python
# wykres słupkowy Seaborn
df = pd.DataFrame({
    "Language": ['Python', 'Java', 'C++', 'Go'],
    "Popularity": [45, 30, 15, 10]
})

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Language", y="Popularity")
plt.title("Popularność języków programowania – Seaborn")
plt.show()
```

```python
# histogram MatPlotLib
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.title("Histogram rozkładu normalnego")
plt.xlabel("Wartość")
plt.ylabel("Częstość")
plt.show()
```

```python
# histogram Seaborn
# Generowanie przykładowych danych
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=500)

df = pd.DataFrame({"value": data})

plt.figure(figsize=(8,5))
sns.histplot(df["value"], kde=True)
plt.title("Histogram z KDE – Seaborn")
plt.xlabel("Wartości")
plt.ylabel("Liczność")
plt.show()
```

```python
# wykres pudełkowy - dane
np.random.seed(0)
df = pd.DataFrame({
    "A": np.random.normal(50, 10, 200),
    "B": np.random.normal(60, 15, 200),
    "C": np.random.normal(55, 7, 200)
})
```

```python
# wykres pudełkowy MatPlotLib
plt.figure(figsize=(8,5))
plt.boxplot([df['A'], df['B'], df['C']], tick_labels=['A','B','C'])
plt.title("Wykres pudełkowy – Matplotlib")
plt.show()
```

```python
# wykres pudełkowy Seaborn
plt.figure(figsize=(8,5))
sns.boxplot(data=df)
plt.title("Wykres pudełkowy – Seaborn")
plt.show()
```

```python
#wykres skrzypcowy Seaborn
plt.figure(figsize=(8,5))
sns.violinplot(data=df)
plt.title("Wykres skrzypcowy – Seaborn")
plt.show()
```

```python
# Scatterplot
tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("Relacja total_bill vs tip")
plt.show()
```

```python
# pairplot
iris = sns.load_dataset("iris")
print(iris.head())
sns.pairplot(iris, hue="species")
```
