### Zadanie 3
Stworzyłem kopię danych z zadania 1.

a) Wypełniłem braki w kolumnie Age medianą.

b) Wypełniłem braki w kolumnie Scholarship wartością 0.

c) Wypełniłem braki w kolumnie Handcap metodą „forward fill”.

d) Wypełniłem braki w AppointmentDay i ScheduledDay metodą „most frequent”.

e) Wyświetliłem DataFrame po imputacji.

```python
# a)
df_3 = df.copy()
df_3['Age'] = df_3['Age'].fillna(df_3['Age'].median())
```

```python
# b)
df_3['Scholarship'] = df_3['Scholarship'].fillna(0)
```

```python
# c)
df_3["Handcap"] = df_3["Handcap"].ffill()
```

```python
# d)
df_3['AppointmentDay'] = df_3['AppointmentDay'].ffill()
df_3['ScheduledDay'] = df_3['ScheduledDay'].ffill()
```

```python
# e)
df_3
```

```python
df_3.info()
df_3.isnull().sum()
```

Porównując z informacjami wyświetlonymi w zadaniu 1 możemy zauważyć, że po zastosowaniu imputacji wyeliminowano braki danych w kolumnach w których dokonywaliśmy zmian (Age, Scholarship, Handcap, ScheduledDay, AppointmentDay).

### Zadanie 5
Wczytaj kopię danych z pliku `data_fill.json`. Przygotuj dane podobnie jak w poprzednich zadaniach (usuń dane, zamień wartości), ale nie wypełniaj danych. Wykorzystując bibliotekę `scikit-learn` zastosuj wielokrotną imputację do wypełnienia brakujących danych numerycznych. Porównaj DataFramy z wypełnionymi danymi.

```python
df_5 = pd.read_json('data_fill.json')
df_5.dropna(inplace=True, subset=['Age', 'Gender', 'Diabetes'])
df_5['Gender'] = df_5['Gender'].map({'M': 0, 'F': 1})
df_5 = df_5.select_dtypes(include=["number"])

imputer = IterativeImputer(random_state=42, max_iter=10)
df_5_imputed = imputer.fit_transform(df_5)
df_5_imputed = pd.DataFrame(df_5_imputed, columns=df_5.columns)
df_5_imputed
```
