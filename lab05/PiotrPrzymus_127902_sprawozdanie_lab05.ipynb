### Zadanie 3
Stworzyłem kopię danych z zadania 1.

a) Wypełniłem braki w kolumnie Age medianą.

b) Wypełniłem braki w kolumnie Scholarship wartością 0.

c) Wypełniłem braki w kolumnie Handcap metodą „forward fill”.

d) Wypełniłem braki w AppointmentDay i ScheduledDay metodą „most frequent”.

e) Wyświetliłem DataFrame po imputacji.

```python
# a)
df_3 = df.copy()
df_3['Age'] = df_3['Age'].fillna(df_3['Age'].median())
```

```python
# b)
df_3['Scholarship'] = df_3['Scholarship'].fillna(0)
```

```python
# c)
df_3["Handcap"] = df_3["Handcap"].ffill()
```

```python
# d)
df_3['AppointmentDay'] = df_3['AppointmentDay'].fillna(df_3['AppointmentDay'].mode()[0])
df_3['ScheduledDay'] = df_3['ScheduledDay'].fillna(df_3['ScheduledDay'].mode()[0])
```

```python
# e)
df_3
```

```python
df_3.info()
df_3.isnull().sum()
```

Porównując z informacjami wyświetlonymi w zadaniu 1 możemy zauważyć, że po zastosowaniu imputacji wyeliminowano braki danych w kolumnach w których dokonywaliśmy zmian (Age, Scholarship, Handcap, ScheduledDay, AppointmentDay).

### Zadanie 5
Wczytałem kopię danych z pliku `data_fill.json`. Przygotowałem dane podobnie jak w poprzednich zadaniach (usunąłem wiersze z brakami danych w kolumnach `Age`, `Gender`, `Diabetes`. Zamieniłem wartości w kolumnie `Gender`. Wykorzystując bibliotekę `scikit-learn` zastosowałem wielokrotną imputację do wypełnienia brakujących danych numerycznych. Potem porównałem DataFramy z wypełnionymi danymi (`df_3` z `df_5_imputed`).

```python
df_5 = pd.read_json('data_fill.json')
df_5.dropna(inplace=True, subset=['Age', 'Gender', 'Diabetes'])
df_5['Gender'] = df_5['Gender'].map({'M': 0, 'F': 1})
df_5 = df_5.select_dtypes(include=["number"])

imputer = IterativeImputer(random_state=42, max_iter=10)
df_5_imputed = imputer.fit_transform(df_5)
df_5_imputed = pd.DataFrame(df_5_imputed, columns=df_5.columns)
df_5_imputed
```

```python
df_3
```

```python
print(df_3.shape)
print(df_5_imputed.shape)
```

W df_5_imputed jest mniej rekordów niż w df_3 z powodu usunięcia wierszy z brakami danych w kolumnach `Age`, `Gender`, `Diabetes` przed użyciem Imputera. W df_3 te wiersze nadal są obecne, ale mają braki danych.

```python
print("----- Różnice w Scholarship -----")
print(df_3['Scholarship'].nunique())
print(df_5_imputed['Scholarship'].nunique())
print("----- Różnice w Hipertension -----")
print(df_3['Hipertension'].nunique())
print(df_5_imputed['Hipertension'].nunique())
print("----- Różnice w Alcoholism -----")
print(df_3['Alcoholism'].nunique())
print(df_5_imputed['Alcoholism'].nunique())
print("----- Różnice w Handcap -----")
print(df_3['Handcap'].nunique())
print(df_5_imputed['Handcap'].nunique())
print("----- Różnice w SMS_received -----")
print(df_3['SMS_received'].nunique())
print(df_5_imputed['SMS_received'].nunique())
```

Podliczyłem liczbę unikalnych wartości w kolumnach `Scholarship`, `Hipertension`, `Alcoholism`, `Handcap` oraz `SMS_received` i zauważyłem, że IterativeImputer wypełnił braki wyliczonymi pośrednimi wartościami. W df_3 te kolumny mają braki danych, ale w df_5_imputed są wypełnione wartościami, które niekoniecznie odpowiadają oryginalnym kategoriom binarnym (poza kolumną `Handcap` która ma 5 unikalnych wartości). IterativeImputer traktuje te kolumny jako numeryczne a tym samym zmienne ciągłe i powoduje utratę logicznej interpretacji wartości w tych kolumnach.

W przypadku takich kolumn lepszym rozwiązaniem byłoby zastosowanie innej metody imputacji, np. zastosowanie modelu klasyfikacyjnego do przewidywania brakujących wartości na podstawie innych cech, oczywiście jest to już bardziej zaawansowane podejście.
