- utworzenie obiektu typu `Series` z 10 losowymi liczbami i własnym indeksem (dla mnie to 127902), wyświetlenie początkowych danych, posortowanie ich rosnąco, i wyświetlenie większych lub równych medianie.

```python
import pandas as pd

# TODO: zadanie 1
import numpy as np

np.random.seed(0)
values = list(np.random.randint(0, 100, size=9)) + [127902]
s = pd.Series(values, index=[f'idx{i}' for i in range(10)])

print("Dane początkowe:")
print(s)
print("\nDane posortowane:")
print(s.sort_values())
print("\nMediana:", s.median())
print("\nElementy >= medianie:")
print(s[s >= s.median()])
```
