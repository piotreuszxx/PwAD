### Zadanie 1
Utworzyłem tablicę zawierającą liczby z przedziału <10;80> z krokiem co 5 za pomocą `np.arrange`, wymiary zmieniłem za pomocą `np.reshape`, transpozycja za pomocą `np.T`. Przy losowych wartościach w drugiej tablicy użyłem stałego seeda żeby wyniki były powtarzalne. Obliczyłem sumę wierszy i kolumn, dodatkowo wyświetliłem je - w celach estetycznych użyłem do tego `Pandas` ze stylami (były mi znane wcześniej). Komórkę w prawym dolnym rogu tabeli, zawierającą zarówno `col_sum` oraz `row_sum` podmieniłem na wartość NaN.

```python
import numpy as np
# a
arr = np.arange(10, 81, 5).reshape(5, 3)
print("arr:\n", arr)

# b
print("Transpozycja:\n", arr.T)

# c
from numpy.random import default_rng
rng = default_rng(42)
rand = rng.integers(0, 100, size=(5, 2))
mat = np.hstack([arr, rand])  # 5x5

# d
row_sum = mat.sum(axis=1)
col_sum = mat.sum(axis=0)
total_sum = row_sum.sum()

import pandas as pd

df = pd.DataFrame(mat, columns=[f'col{i}' for i in range(mat.shape[1])])
df['row_sum'] = row_sum
df.loc['col_sum'] = list(col_sum) + [total_sum]

df.loc['col_sum', 'row_sum'] = np.nan

styled = (
    df.style
      .format("{:.2f}", na_rep="")
      .apply(lambda row: ['font-weight: bold; background-color: gray; color: black' if row.name == 'col_sum' else '' for _ in row], axis=1)
      .apply(lambda col: ['font-weight: bold; background-color: gray; color: black' if col.name == 'row_sum' else '' for _ in col], axis=0)
)

styled
```

### Zadanie 2
Wczytałem dane z pliku `data_numpy.csv` i wykonałem zadania a-e. W podpunkcie d) ponownie w celach estetycznych zastosowałem `Pandas` ze stylami.

#### Kod źródłowy

```python
# a
data = np.loadtxt('data/data_numpy.csv', delimiter=',')
print('shape:', data.shape)
```

```python
# b
col_min = data.min(axis=0)
col_max = data.max(axis=0)
print('col_min:', col_min)
print('col_max:', col_max)
```

```python
# c
row_mean = data.mean(axis=1)
row_std = data.std(axis=1)
print('row_mean:', row_mean)
print('row_std:', row_std)
```

```python
# d
import pandas as pd

trans = data.T
df = pd.DataFrame(trans, columns=[f'col{i}' for i in range(trans.shape[1])])
df.loc['mean'] = row_mean
df.loc['std'] = row_std
trans_with_stats = df.values

print('po transpozycji z dodanymi statystykami:', trans_with_stats.shape)
styled = (
    df.style
      .format('{:.2f}')
      .apply(lambda row: ['font-weight: bold; background-color: gray; color: black' if row.name in ('mean','std') else '' for _ in row], axis=1)
)

styled
```

```python
# e
np.savetxt('data/results_numpy_task2.csv', trans_with_stats, fmt='%.2f', delimiter=',')
print("Zapisano do data/results_numpy_task2.csv")
```

### Zadanie 3

$$
\frac{x-min}{max-min}\
$$
używając `col_min` oraz `col_max` z zadania 2b.

#### Kod źródłowy

```python
normalized = (data - col_min) / (col_max - col_min)
print('Znormalizowane dane:\n', normalized[:5])
```
