### Przykłady

```python
# Tworzenie obiektu ndarray i podstawowe operacje
arr = np.array([[2.5, -.5, 3.1], [.5, -3, 7.5]])
print(arr)
print(arr * 10)
print(arr + arr)
print(arr - arr)
print(arr / 2)
```

```python
# Atrybuty i agregacje
print(f"ndim: {arr.ndim}")
print(f"shape: {arr.shape}")
print(f"dtype: {arr.dtype}")
print(f"size: {arr.size}")


print(f"min: {arr.min()} at {arr.argmin()}")
print(f"max: {arr.max()} at {arr.argmax()}")
print(f"mean: {arr.mean()}")
print(f"sum: {arr.sum()}")
print(f"prod: {arr.prod()}")
print(f"std: {arr.std()}")
print(f"var: {arr.var()}")
```

```python
print(arr)
print(arr.transpose())
```
