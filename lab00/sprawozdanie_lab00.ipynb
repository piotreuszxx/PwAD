## Zadanie 1
- a) uzupełnienie funkcji `even_sum(l)` - zwraca sumę parzystych elementów listy.
- b) testy jednostkowe z użyciem `unittest`.

```python
def even_sum(l):
    sum = 0
    for i in l:
        if i % 2 == 0:
            sum += i
    return sum
    pass

print(even_sum([1,2,3,4,5,6]))  # powinno zwrócić 12
```

```python
import unittest

class TestEvenSum(unittest.TestCase):
    def test_empty_list(self):
        self.assertEqual(even_sum([]), 0)

    def test_all_odd(self):
        self.assertEqual(even_sum([1, 3, 5]), 0)

    def test_all_even(self):
        self.assertEqual(even_sum([2, 4, 6]), 12)

    def test_mixed(self):
        self.assertEqual(even_sum([1, 2, 3, 4, 5]), 6)

if __name__ == '__main__':
    unittest.main(argv=[''], verbosity=2, exit=False)
```

## Zadanie 2
- a) uzupełniłem klasę `Student` o atrybuty `name` (już istniał w nadklasie Person), `surname`, `ects_points`.
- b) dodałem metodę `add_points` z walidacją (ValueError dla wartości ujemnej).
- c) zaimplementowałem `__str__` zwracający czytelną reprezentację obiektu.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, surname):
        super().__init__(name, None)
        self.surname = surname
        self.ects_points = 0
        pass

    def add_points(self, points):
        if points < 0:
            raise ValueError("ECTS points cannot be negative.")
        self.ects_points += points

    def __str__(self):
        return f"Student: {self.name} {self.surname}, ECTS points: {self.ects_points}"
        pass

s = Student("Jan", "Kowalski")
s.add_points(10)
print(s)
```

## Zadanie 3
Napisałem skrypt otwierający plik `data03.txt`, liczący wszystkie wiersze i liczbę pustych wierszy. Zapisuje on wyniki do pliku `results03.txt`.
Skrypt zgodnie z poleceniem zawiera obsługę wyjątków.

```python
import re

try:
    with open('data03.txt', 'r') as file:
        lines = file.readlines()
        total_lines = len(lines)
        empty_lines = sum(1 for line in lines if re.match(r'^\s*$', line))

    with open('results03.txt', 'w') as results_file:
        results_file.write(f'Total lines: {total_lines}\n')

    print(f'Number of empty lines: {empty_lines}')

except FileNotFoundError:
    print("File data03.txt not found.")
except IOError as e:
    print(f'An I/O error occurred: {e}')
```
