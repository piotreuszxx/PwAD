### Przykłady

```python
# Funkcja licząca n-ty element ciągu Fibonacciego
def fibonacci(n: int) -> int:
    if n < 0:
        raise ValueError("n can not be negative")
    elif n == 0:
        return 0
    elif n == 1:
        return 1

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b


# przykład użycia
n = int(input("n =  "))
print(f"{n}th element of the sequence: {fibonacci(n)}")
```

```python
# Klasa z metodą i obsługą błędu
class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def payment(self, amount):
        if amount < 0:
            raise ValueError('The amount cannot be negative!')
        self.balance += amount

acc = Account()
acc.payment(100)
print(acc.balance)
```

#### Kod źródłowy

```python
def even_sum(l):
    # TODO: a)
    sum = 0
    for i in l:
        if i % 2 == 0:
            sum += i
    return sum
    pass

print(even_sum([1,2,3,4,5,6]))  # powinno zwrócić 12
```

#### Testy jednostkowe

```python
# TODO: b)
import unittest

class TestEvenSum(unittest.TestCase):
    def test_empty_list(self):
        self.assertEqual(even_sum([]), 0)

    def test_all_odd(self):
        self.assertEqual(even_sum([1, 3, 5]), 0)

    def test_all_even(self):
        self.assertEqual(even_sum([2, 4, 6]), 12)

    def test_mixed(self):
        self.assertEqual(even_sum([1, 2, 3, 4, 5]), 6)

if __name__ == '__main__':
    unittest.main(argv=[''], verbosity=2, exit=False)
```
